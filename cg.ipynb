```python
import numpy as np
import math
import rogues as rg

import matplotlib.pyplot as plt 
```

```python
def cg(Av,x,b,tol=1.0e-9):
    n = len(b)
    r = b - Av(x)
    s = r.copy()
    y = np.copy(x)
    for i in range(n):
        u = Av(s)
        alpha = np.dot(s.T,r)/np.dot(s.T,u)
        x = x + alpha*s
        r = b - Av(x)
        y = np.hstack([y,x])
        if(math.sqrt(np.dot(r.T,r))) < tol:
            break
        else:
            beta = -np.dot(r.T,u)/np.dot(s.T,u)
            s = r + beta*s
    return x, i, y
```

```python
A=np.array([[3.,2.],[2.,6.]])    
b=np.array([[5.],[8.]])
x=np.array([[0.],[.0]])

def Av(x):
    return A@x
    
x,k,y = cg(Av,x,b,tol=1.0e-9)

print("sol=",x)
print("iter=",k)
```

```python
plt.plot(y[0,:],y[1,:],'*-');plt.axis('equal');plt.grid();
```
