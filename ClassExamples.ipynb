## - Inheritance

```python
class Pet():

  def __init__(self, name="Rex", age=2):
    #this is the constructor
    self.name = name
    self.age = age
  
  def Jump(self):
    #this is a method
    print(self.name, ' jumped!')

```

```python
# "a" is a Pet object
a = Pet("Laica", 5)

print(a)
```

```python
# "a" has properties name and age
print(a.name)
print(a.age)
```

```python
# "a" has the method Jump()
a.Jump()
```

```python
class Dog(Pet):
  #Dog has the same constructor and the same Jump() method as his parent

  def MakeSound(self):
    print(self.name , "said wouf")

  
```

```python
b = Dog()
b.MakeSound()
b.Jump()
```

```python
class Dog2(Pet):
  #Dog2 has the same constructor as his parent

  #Dog2 has the same MakeSound as his parent

  #Dog2 has a different method Jump() as his parent
  def Jump(self):
    print(self.name , "doesn't want to jump")
```

```python
c = Dog2()
c.Jump()
```

```python
class Dog3(Pet):
  #Dog3 has the same constructor as his parent

  #Dog2 has the same MakeSound as his parent

  #Dog3 has the same method Jump() as his parent, plus extra functionalities
  def Jump(self):
    super().Jump()
    print(self.name , "is too tired to jump again :(")
```

```python
d = Dog3()
d.Jump()
    
```

```python
class GeometricShape():

  def __init__(self, colour="red"):
    self.colour = colour
```

```python
a = GeometricShape()
```

```python
print(a.colour)
```

```python
class Square(GeometricShape):

  def CalculateArea(self, side=5):
    print("The area of the ", self.colour, " square is", side*side)
```

```python
b = Square()
```

```python
b.CalculateArea()
```
